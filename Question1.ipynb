```python
# Install required libraries
!pip install -q transformers torch

from transformers import pipeline
import torch

# Initialize generator (auto-uses GPU if available)
generator = pipeline(
    'text-generation',
    model='gpt2-medium',  # Better quality than base GPT-2
    device=0 if torch.cuda.is_available() else -1,
)

# Personal introduction prompt (Descriptive)
descriptive_prompt = """
My name is Hari. I am from TamilNadu.
I am studying at Anna University as a second-year CSE student.
"""

# Directive prompt (explicit instructions)
directive_prompt = """
Write a short paragraph about Prakash with these rules:
1. Start with his name and hometown.
2. Mention his university and degree.
3. Keep the tone formal and concise.
4. Use exactly 3 sentences.
Text:
"""

# Generation function
def generate_text(prompt):
    output = generator(
        prompt,
        max_length=150,
        temperature=0.7,  # Balanced creativity
        top_p=0.9,
        do_sample=True,
        num_return_sequences=1,
    )
    return output[0]['generated_text']

# Run experiments
print("=== DESCRIPTIVE PROMPT (Natural) ===")
print(generate_text(descriptive_prompt))

print("\n=== DIRECTIVE PROMPT (Structured) ===")
print(generate_text(directive_prompt))
```
